Given the Candy domain and the Box<T> class:

```java
public class Candy {
    public static final int FRESH_FACTOR = 6;

    public boolean isStale() { return new Random().nextInt(FRESH_FACTOR) % FRESH_FACTOR == 0; }
}

public class Chocolate extends Candy {}

public class Box<T> {
    private List<T> values = new ArrayList<>();

    public void add(T t){
        this.values.add(t);
    }

    public boolean isFull() {
        return values.size() > 0;
    }

    public List<T> getValues() { return values; }
}
```

And here are your candies 🍬 and chocolates 🍫:

```java
Box<Candy> candies = new Box<>();
Box<Chocolate> chocolates = new Box<>();
```

Also given this inspect method:

```java
void inspect(Box<Candy> box) {
    for(Candy c : box.getValues()) {
        if(c.isStale()) throw new RuntimeException("I'm not a fool!");
    }
    if(!box.isFull()) {
        box.add(new Chocolate());
    }
}
```

We can inspect candies:

```java
inspect(candies);
candies.isFull(); // true
```
